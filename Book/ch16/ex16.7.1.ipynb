### Ex 16.7.1 - Prove ATET is id:d by DiD
We wish to show that under the assumption that potential outcomes satisfy conditional parallel trends
$$ \mathbb{E}\!\left[Y_2(0)-Y_1(0)\mid D=1,\,X\right]
\;=\;
\mathbb{E}\!\left[Y_2(0)-Y_1(0)\mid D=0,\,X\right]\ \text{a.s.} $$

and no anticipation 
$$
\mathbb{E}\!\left[Y_1(0)\mid D=1,\,X\right]
\;=\;
\mathbb{E}\!\left[Y_1(1)\mid D=1,\,X\right]\ \text{a.s.}
$$

with a control group, with overlapping charctersics 
$$
\exists\,\varepsilon>0:\quad \mathbb{P}(D=1)\ge \varepsilon
\quad\text{and}\quad \mathbb{P}(D=1\mid X)\le 1-\varepsilon\ \text{a.s.}
$$

The ATET we wish to recover is defined

$$
\alpha := \mathbb{E}\!\left[\,Y_2(1)-Y_2(0)\mid D=1\,\right].
$$

We can decompose the ATET
$$
\begin{aligned}
\mathbb{E}\!\left[Y_2(1)-Y_2(0)\mid D=1,X\right]
&= \mathbb{E}\!\left[\big(Y_2(1)-Y_1(1)\big)\mid D=1,X\right] \\
&\quad - \mathbb{E}\!\left[\big(Y_2(0)-Y_1(0)\big)\mid D=1,X\right] \\
&\quad + \mathbb{E}\!\left[Y_1(1)-Y_1(0)\mid D=1,X\right].
\end{aligned}
$$

By no anticipation:

$$
\mathbb{E}\!\left[Y_1(1)-Y_1(0)\mid D=1,X\right] = 0.
$$

Using parallel trends:

$$
\mathbb{E}\!\left[Y_2(0)-Y_1(0)\mid D=1,X\right]
=
\mathbb{E}\!\left[Y_2(0)-Y_1(0)\mid D=0,X\right].
$$

Thus conditional ATET:

$$
\alpha(X)
=
\mathbb{E}\!\left[\,\Delta Y \mid D=1,X\right]
-
\mathbb{E}\!\left[\,\Delta Y \mid D=0,X\right],
\quad \Delta Y := Y_2-Y_1.
$$

Finally, average over the treated distribution of \(X\):

$$
\alpha
=
\mathbb{E}\!\Big[
\mathbb{E}\!\left[\Delta Y \mid D=1,X\right]
-
\mathbb{E}\!\left[\Delta Y \mid D=0,X\right]
\ \Big|\ D=1
\Big].
$$
